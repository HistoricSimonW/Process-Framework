```python
from process_framework.pipeline import PipelineDefinitionBase, SettingsBase, ClientsBase, ReferencesBase
from process_framework.pipeline.references import ReferencesDefinitionBase
from process_framework.pipeline.cli import CliArg
from process_framework.steps import Step
from process_framework.steps.logging_step import LogMessage, LogInput
from process_framework import Reference
from dataclasses import dataclass


class PipelineBuilder(PipelineDefinitionBase):
    def instantiate_steps(self) -> list[Step]:
        self.references.message = self.settings.message
        return [
            LogMessage("running!"),
            LogInput(input_=self.references.message)
        ]


class Settings(SettingsBase):
    message:str = CliArg('--message')


class Clients(ClientsBase):
    ...


@dataclass
class References(ReferencesBase):
    message:Reference[str]

        
class ReferencesDefinition(ReferencesDefinitionBase[References]):

    @classmethod
    def instantiate(cls) -> References:
        return References(
            message=Reference(str)
        )
```

```python
from process_framework.pipeline.cli import Runner

cli = Runner(
    definition=PipelineBuilder,
    settings_type=Settings,
    clients_type=Clients,
    references_type=ReferencesDefinition
)

cli.main(args)
```
