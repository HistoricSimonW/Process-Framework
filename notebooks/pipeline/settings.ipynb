```python
class Settings(ElasticsearchIndexArgs, ElasticsearchIngestPipelineArgs, ElasticsearchClientArgs, EngineArgs, SettingsBase):
    ...
```

```python
from pathlib import Path
class Databases(BaseModel):
    a:EngineArgs
    b:EngineArgs


class Paths(BaseModel):
    renames:Path
    cla_gr_to_concept_scheme:Path


class NestedSettings(Settings):
    dbs:Databases
    paths:Paths
    a:str|None
    b:str|None


NestedSettings(
    a='',
    b='',
    host='',
    database='',
    username='',
    password='',
    drivername='',
    query={'driver':'a'},
    ingest_pipeline='',
    index='',
    cloud_id='',
    api_key='',
    dbs=Databases(
        a=EngineArgs(
            host='',
            database='',
            username='',
            password='',
            drivername='',
            query={'driver':'a'},
        ),
        b=EngineArgs(
            host='',
            database='',
            username='',
            password='',
            drivername='',
            query={'driver':'a', 'trust_server_certificate':'b'},
        )
    ),
    paths=Paths(
        renames='../parent/file.json',
        cla_gr_to_concept_scheme='../parent/folder/file.json'
    )
)
```

```python
s = NestedSettings(_env_file='.env.demo')
s.get_client()
```
