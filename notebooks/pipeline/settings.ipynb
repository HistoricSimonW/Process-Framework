```python
# from process_framework.pipeline import SettingsBase
from process_framework.pipeline.composition.core import ElasticsearchIndexArgs, ElasticsearchIngestPipelineArgs, DatabaseArgs
from pydantic import BaseModel
from pydantic_settings import SettingsConfigDict, BaseSettings
```

```python
class Settings(ElasticsearchIndexArgs, ElasticsearchIngestPipelineArgs, DatabaseArgs, BaseSettings):
    model_config = SettingsConfigDict(
        env_file=".env",
        env_nested_delimiter="__",
        case_sensitive=False
    )
```

```python
from pathlib import Path
class Databases(BaseModel):
    a:DatabaseArgs
    b:DatabaseArgs


class Paths(BaseModel):
    renames:Path
    cla_gr_to_concept_scheme:Path


class NestedSettings(Settings):
    dbs:Databases
    paths:Paths


NestedSettings(
    host='',
    database='',
    username='',
    password='',
    drivername='',
    query_driver='',
    ingest_pipeline='',
    index='',
    dbs=Databases(
        a=DatabaseArgs(
            host='',
            database='',
            username='',
            password='',
            drivername='',
            query_driver='',
        ),
        b=DatabaseArgs(
            host='',
            database='',
            username='',
            password='',
            drivername='',
            query_driver='',
        )
    ),
    paths=Paths(
        renames='../parent/file.json',
        cla_gr_to_concept_scheme='../parent/folder/file.json'
    )
)
```

```python
NestedSettings(_env_file='.env.demo')
```
