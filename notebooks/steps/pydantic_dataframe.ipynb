```python
import pandas as pd
from pandas import DataFrame
from pydantic import BaseModel
```

```python
class Car(BaseModel):
    name:str
    horsepower:int
    color:str


cars = [
    Car(name='volvo', horsepower=200, color='white'),
    Car(name='golf', horsepower=100, color='red')
]

DataFrame.from_records([car.model_dump() for car in cars])
```
