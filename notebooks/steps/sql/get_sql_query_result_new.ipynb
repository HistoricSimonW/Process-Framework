```python
from process_framework.steps.mixins.sql import (
    IGetQueryResult, IGetQuery, ILimitQuery, IAddInClause,
    GetTextQueryResult, AssignTextQueryResult
)
from process_framework.references import Reference
```

```python
from sqlalchemy import text, select
from dataclasses import dataclass

@dataclass
class GetCars(IGetQueryResult, IGetQuery):
    def get_query(self):
        return select(text("select top 10 * from cars"))
    
vs = Reference(list, ['1', '2', '3'])
gc = GetCars(
    engine='',
    index_col='',
    dtype='',
    renames=dict(),
    modifiers=[
        IAddInClause(
            in_values=vs,
            in_column='VIN',
        ),
        ILimitQuery(
            count=20
        )
    ]
)

str(gc.get_modified_query())
```

```python
gc = GetTextQueryResult(
    modifiers=[
        IAddInClause(
            in_values=vs,
            in_column='VIN',
        ),
        ILimitQuery(
            count=20
        )
    ], 
    query='select top 10 * from cars"'
)

str(gc.get_modified_query())
```
