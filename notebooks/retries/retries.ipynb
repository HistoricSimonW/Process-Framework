```python
import random
from tenacity import retry
```

```python
@retry
def do_something_unreliable():
    if random.randint(0, 10) > 1:
        raise IOError('OOPS!')
    else:
        return True
    
print(do_something_unreliable())
```

```python
from tenacity import stop_after_attempt, wait_fixed
from time import sleep

@retry(wait=wait_fixed(1), stop=stop_after_attempt(3), retry_error_callback=print)
def do_something_unreliable():
    if random.randint(0, 10) < 9:
        raise IOError('OOPS!')
    else:
        return True
    
print(do_something_unreliable())
```

```python
from tenacity import stop_after_attempt, wait_exponential
from time import sleep

@retry(wait=wait_exponential(2), stop=stop_after_attempt(3), retry_error_callback=print, before=print)
def do_something_unreliable():
    if random.randint(0, 10) < 9:
        raise IOError('OOPS!')
    else:
        return True
    
print(do_something_unreliable())
```
